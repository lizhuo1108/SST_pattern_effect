```python
species = ("TTEND_TOT'", "H'", "V'","DTCOND'","QRS'","QRL'","DTV'","RES")
penguin_means = {
    'W for Med_JN': mean_list_Med,
    'W for Car_JN': mean_list_Car,
    'C for Ind_JN': mean_list_Ind,
}

x = np.arange(len(species))  # the label locations
width = 0.25  # the width of the bars
multiplier = 0

fig = plt.figure(figsize=(6, 4),dpi=600,constrained_layout=True)
grid = fig.add_gridspec(ncols=1,
                        nrows=1)
ax = fig.add_subplot(grid[0, 0])
color_list=['blue','red','cyan']
for attribute, measurement in penguin_means.items():
    offset = width * multiplier
    rects = ax.bar(x + offset, measurement, width, color=color_list[multiplier],label=attribute)
#     ax.bar_label(rects, padding=3)
    multiplier += 1
ax.axhline(y=0,color='k',linestyle='-',linewidth=1)    
ax.set_xticks(x + width, species)
ax.legend(loc='best')
ax.set_ylim([-0.8,1.25])
ax.set_title(r'$\theta$ budget diff (K/d)',pad=1)
```
